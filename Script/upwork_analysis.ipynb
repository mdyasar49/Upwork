# Upwork Lead Scraping & Opportunity Analytics

This notebook provides interactive exploration, data cleaning, and visualization for leads scraped from Upwork and Freelancer.

```python
import pandas as pd
import matplotlib.pyplot as plt
import re
from pathlib import Path

output_csv = Path("../Output/upwork_job_leads.csv")
if output_csv.exists():
    df = pd.read_csv(output_csv)
    print(f"Total leads loaded: {len(df)}")
    display(df.head(10))
else:
    print("No scraped CSV found in Output/. Run upwork_scraper.py first.")
```

```python
# Filter leads with verified Contact Details (Email or Phone)
if 'df' in locals() and not df.empty:
    contacts_df = df[(df['Email_Address'].fillna('').str.len() > 0) | (df['Contact_Number'].fillna('').str.len() > 0)]
    print(f"Leads with Direct Contact Info: {len(contacts_df)}")
    display(contacts_df[['Job_Title', 'Contact_Number', 'Email_Address', 'Website_URLs', 'Client_Location']])
```

```python
# Top In-Demand Skills
if 'df' in locals() and not df.empty:
    all_skills = df['Skills'].dropna().str.split(',').explode().str.strip()
    all_skills = all_skills[all_skills.str.len() > 0]
    top_skills = all_skills.value_counts().head(15)
    print("Top Skills in demand:")
    print(top_skills)
    top_skills.plot(kind='barh', title='Top 15 Skills Required in Scraped Jobs', color='teal')
    plt.gca().invert_yaxis()
    plt.show()
```
